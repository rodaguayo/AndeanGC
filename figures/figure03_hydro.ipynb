# Hydrological signatures figure


```python
import pandas as pd
import numpy as np

from plotly.subplots import make_subplots
import plotly.graph_objects as go

from andeangc import config as cfg
```

```python
data = pd.read_csv(cfg.VERSION / 'AndeanGC_signatures.csv', index_col=0)

# Gauges below cfg.sig_min_years are NaN in every signature but sig_n_years, so only the
# first panel counts all of them.
signatures = {'sig_n_years':    'Complete Years',
              'q_mean':         'Mean Flow (mm/d)',
              'q5':             'Q5 (mm/d)',
              'q95':            'Q95 (mm/d)',
              'runoff_ratio':   'Runoff Ratio',
              'stream_elas':    'Streamflow Elast.',
              'slope_fdc':      'FDC Slope',
              'baseflow_index': 'Baseflow Index',
              'recession_k':    'Recession k (1/d)',
              'hfd_mean':       'Half-Flow Date',
              'q_max_doy':      'Max-Flow Date',
              'q_min_doy':      'Min-Flow Date',
              'q_centroid_doy': 'Flow Centroid',
              'seasonality':    'Seasonality',
              'high_q_freq':    'High-Flow Freq. (d/yr)',
              'high_q_dur':     'High-Flow Dur. (d)',
              'low_q_freq':     'Low-Flow Freq. (d/yr)',
              'low_q_dur':      'Low-Flow Dur. (d)'}
```

## Plot

```python
# Calculate subplot dimensions
n_rows = (len(signatures) + 5) //5  # 5 columns per row

# Create subplots
fig = make_subplots(
    rows=n_rows,
    cols=5,
    subplot_titles=list(signatures.values()),
    vertical_spacing=0.07,
    horizontal_spacing=0.03,
)

fig.for_each_annotation(lambda a: a.update(font=dict(size=12)))

# Create histogram for each signature
for i, attr in enumerate(signatures):
    row = i // 5 + 1
    col_idx = i % 5 + 1

    fig.add_trace(
        go.Histogram(x=data[attr], name=attr, showlegend=False, nbinsx=30, marker_color="#3979a7"),
        row=row,
        col=col_idx,
    )

# Update axes styling
fig.update_xaxes(griddash="dot", ticks="outside", title_standoff=5, tickfont=dict(size=10))
fig.update_yaxes(griddash="dot", ticks="outside", title_standoff=5, tickfont=dict(size=10))

# Update layout
fig.update_layout(
    height=700,
    width=1000,
    template="seaborn",
    margin=dict(l=10, r=10, b=10, t=25)  # three tall rows push the row-1 titles above the paper edge
)

fig.show()
fig.write_image(cfg.VERSION / "figures" / 'figure03_hydro.png', scale=4, height=700, width=1000)
```
